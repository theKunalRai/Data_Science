#### Differemt types of operators
- What are operators?
- What are the different types of operators?
- Comparison, Arithmatic, Logical, Bitwise, Assignment operators

```python
# Arithmatic Operators
# +,-,/,*,//,% and exponential
a = 10
b = 4
print(a+b)
print(a-b)
print(a*b)
print(a/b) # gives exact float value of division
print(a//b) #gives quotient
print(a%b) #gives remainder
# 2 raised to the power of 3, 2*2*2
# let's do 'a' raised to the power of 'b'
print(a**b)
```

### Comparison operators
- ==,!=,<=,>=,<,>

```python
x = 5
y = 6
print(x,y)
```

```python
print("Is x equal to y:",x==y) #check if x is equal to y
print("Is x not equal to y:",x!=y) #check if x is not equal to y
print("Is x less than or equal to y:",x<=y) # checks if x is less than or equal to y
print("Is x greater than or equal to y:",x>=y) # checks if x is greater than or equal to y
print("Is x less than y:",x<y) # checks if x is less than y
print("Is x greater than y:",x>y) # checks if x is greater than y
```

### Assignment Operators

```python
print(x,y)
```

```python
x += y # x becomes x+y
print(x,y)
```

```python
x -= y # x becomes x-y, it takes the last value of x
print(x,y)
```

```python
x *= y # x becomes x*y
print(x,y)
```

```python
x /= y # x becomes x/y
print(x,y)
```

```python
x **= y # x becomes x^y
print(x,y)
```

```python
x //= y # x becomes quotient when divided by y
print(x,y)
```

```python
x %= y 
print(x,y)
```

### Different Data Structures
- Lists, Tuples, Sets and Dictionaries

```python
# on the first day, three peope have enrolled into the course
# Rahees, Ram, and Robert
# where in the management wants to store these names into some variable
names = ['Rahees','Ram','Robert'] #list is created using square brackets, we can also use list() fn to do the same
print("The list of the people enrolled in the course so far:",names)
print(type(names))
```

```python
names = list(['Rahees','Ram','Robert'])
print("List of people enrolled:",names)
```

```python
# lets explore the functions that are available on lists
# on the second day rohan took the admission, now we have to add student tot the list
names.append(['Rohan'])
print("List of People Enrolled so far",names)
```

```python
names.append(['Ranjan','Riaz'])
print("list of people enrolled so far",names)
```

```python
# lets remove the very last elements to the list
names.pop()
print(names)
```

```python
names.pop()
print(names)
```

```python
names.extend(['Riaz'])
print(names)
```

```python
names.extend(['Ronald','Ranjan'])
print(names)
```

```python
names.extend(['Randheer','Ranjeet'])
print(names)
```

```python
# writye  auser defined fn to find the product of 3 number given
# parameters and arguments
def product(a,b,c): # here a,b,c act like placeholders we call them parameters
    return a*b*c
product(2,4,5) #the values 2,4,5 are called arguments
```

```python
# write a function that takes a number as input and returns it's square and square_root
def sq_sqrt(a):
    print("The square of a given number is: ",a**2)
    print("The square root of a given number is: ",a**0.5)
sq_sqrt(25)
```

```python
# write a user defined fn in python that can find the absolute value of a number given
def abs(num):
    if num<0:
        print("Absolute value of the given number is: ",num*-1)
    else:
        print("Absolute value of the given number is: ",num)

abs(-45)
```

```python
#write a udf that classify a given number into even or odd
def even_odd(num):
    if num%2==0:
        print("it's an even no.")
    else:
        print("it's an odd no.")

even_odd(44)
```

```python
# create a udf that takes string as an input and returns it's lower, capital, upper and title versions.
def case_version(n):
    return n.lower(),n.upper(),n.capitalize(),n.title()
case_version("KUnal rAi")
```

### while condition:
    # the code has to repeat
- if the condition is it would execute the code
- after execution, check the condition again
- if the condition becomes false, stop the loop

```python
i = 1
while i<=6:
    print(i)
    i=i+1
```

```python
i = 0
while i<=9:
    print(i)
    i=i+1
```

# iteration in the above prgm

- step 2, i=2, it checks the condition i<=6, print(2)
- step 5, i=5, it checks the condition i<=6, print(5)
- step 6, i=6, it checks the condition i<=6, print(6)
- step 7, i=7, it checks the condition i<=6, condition fails, False, then loop end

```python
# fundamental difference between for loop and while loop
# for loop is used when we genrally know what we are iterating through
for i in range(5):
    print(i)
```

```python
# while loop is used when you want to continue until some changes in the condition happens
i = 1
while i<=5:
    print(i+2)
    i+= 1
```

- i = 1, 3
- i = 1+1, 4
- i = 2+1, 5
- i = 3+1, 6
- i = 4+1, 7
- i = 5+1, false, loop ends here

```python
# using break
# break terminates the loop
i =1
while i<=10:
    print(i)
    if i==5:
        break
    i += 1
```

- even though i<=10 is still true, break stops the loop

```python
# using continue 
# conitnue skips the current iteration and moves to the next one 
i = 0
while i<5:
    i += 1
    if i==3:
        continue 
    print(i)
```

- when i becomes 3, continue skips the print() statement and goes to next iteration

### lambda functions

```python
# consider finding the product of two numbers
def prod(a,b):
    return a*b
prod(3,4)
```

- prod is the fn name
- a,b are called as parameters
- 3 and 4 are called as arguments
- a*b is the value returned by the fn

```python
sum_2 = lambda a,b:a+b
sum_2(2,4)
```

```python
print((lambda a,b:a*b)(3,4))
```

```python
# write a lambda fn to find square and a cube of a number
sq = lambda x:x**2
sq(4)
```

```python
# write a lambda fn to find square and a cube of a number
print((lambda x:x**2)(6))
```

```python
- we have to use the keyword "lambda"
- pass as many parameters as you want
- but there should be only one return statement
```

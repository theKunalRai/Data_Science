```python
# we have to create 4 states and their capitals like a pair
# "Goa","Panaji"
# "Tamil Nadu","Chennai"
# "Telangana","Hyderabad"
# "Bihar","Patna"
# when we want to create an object with the elements in the form of pairs, we use dictionaries
states_capitals = {"Goa":"Panaji",
                  "Telangana":"Hyderabad",
                  "Tamil Nadu":"Chennai",
                  "Bihar":"Patna"
                  }
print(states_capitals)
print(type(states_capitals))
```

```python
# let's fetch all the items/elements of the dictionary
states_capitals.items()
```

```python
# let's fetch all the keys
states_capitals.keys()
```

```python
states_capitals.values() # fetch all the values
```

```python
len(states_capitals)
```

```python
states_capitals.get("Telangana")
```

- items() fn fetches all the elements in the dictionary
- keys() fn fetches all the keys
- values() fn fetches all the values assigned to all the keys
- get() fn fetches 'value' of a particular 'key'

```python
# now add "Kerala" and "Thiruvananthpuram" to the dictionary
states_capitals.update({"Keralam":"Thiruvananthpuram"})
print(states_capitals)
```

```python
# let's add another one "Assam" and its capital "Dispur"
states_capitals["Assam"]="Dispur"
print(states_capitals)
```

- we can add new items to the dictionary in two ways
- using update() fn and the above one

```python
states_capitals.pop("Assam")
print(states_capitals)
```

```python
states_capitals.pop('Keralam')
print(states_capitals)
```

```python
states_capitals.popitem() # popitme() fn is used to delete recently added element
print(states_capitals)
```

### for loops 

```python
# Let's create a list of first natural numbers
list_10 = [x for x in range(1,11)]
print(list_10)
```

```python
# now take every element of list in loop and print the square of it
for x in list_10:
    print(x**2)
```

```python
# Let's create an empty list of square and append the squares of elements into it
# empty list
squares = []
for i in list_10:
    squares .append(i**2)
print(squares)
```

```python
# Let's create an empty tuple of square and append the squares of elements into it
# empty list
squares_tuple = ()
for i in list_10:
    squares_tuple.append(i**2)
print(squares_tuple)
```

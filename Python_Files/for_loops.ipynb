```python
# let's create a list of first 20 whole numbers
list_20 = [x for x in range(20)]
print(list_20)
```

```python
# create a list of salaries with five elements in it
salary = [25000,35000,200000,600000,300000]
# create a list savings with five elements
saving = [12000,40000,30000,500000,80000]
incentives = [1000,2000,4000,6000,7000]
# using for loop create a list of expenditure, salary - savings = expenditure
expenditure=[] # expenditure = salary - saving + incentives
for i,j,k in zip(salary,saving,incentives):
    expenditure.append(i-j+k)
print(expenditure)
```

```python
# create a list of 4 states and give the names of four states with the mix of upper and/or lower cases
states = ['uTTaRaKHaND','pUnjAb','HaRYaNa','KaRNaTaKa','uttar pradesh','go goa gone']
upper_states=[]
lower_states=[]
capitalized_states=[]
titled=[]
for i in states:
    lower_states.append(i.lower()),
    upper_states.append(i.upper()),
    capitalized_states.append(i.capitalize())
    titled.append(i.title()) # makes every word's first letter in statement or a name, in caps, whereas capitalize() only makes the first word caps.
print(upper_states)
print(lower_states)
print(capitalized_states)
print(titled)
```

### Creating Variables

- First let's create string type of variables
- name, qualification, designation, college
- While creating string type of variables, we can use single quotes, double quotes or even triple quotes
- Usually triple quotes are used when we are assigning long paragraphs to a variable.

```python
# let's create two variables name and qualification
name = 'Kunal Rai'
qualification = 'BE in Electronics and Communication'
print(name)
print(qualification)
```

```python
# lets check the data type of the variables name and qualification
# in order to check the type of the variable, we can use type()
# here type() is a pre-defined function
# a function is always followed by a paranthesis
# within the fucntion we write the variable-name
print(type(name))
print(type(qualification))
```

```python
# let's print the variable in a little explanatory way
print("My Name is",name)
print("My Qualification is",qualification)
```

```python
# let's create some numeric variables age and salary
age = 31
salary = 2445678.34
print(type(age))
print(type(salary))
print("My age is",age)
print("My Salary is",salary)
```

### let's understand the rules for naming the variables
- a variable name always start with an alphabet

- a variable name can contain alphabets and numerics
- cannot start with a numeric but can be there in anyh other position
- other than underscore , not other special characters are allowed to be a part of variable name
- there should not be a space in between names of the variables
- keywords can not be a variable name

### Different types of operators in python
- Arithmetic
- Logical Operators
- Bitwise Operators

```python
# lets create two integer type variables and name them a and b
a = 9
b = 5
# sum of a and b
print("sum of a and b is:",a+b)
# subtraction of a and b
print("subtraction of a and b is:",a-b)
#product of a and b
print("product of a and b is:",a*b)
#division of a by b
print("division of a by b is:",a/b) # gives exact division 
print("quotient of a divided by b is:",a//b) #gives quotient of division of a by b
print("remainder we get in the division of a by b is:",a%b) # gives remainder of division of a by b
```

### List Comprehensions

```python
list_10 = [x for x in range(15)]
print(list_10)
```

```python
# Let's create a list of squares of first 14 whole numbers
list_squares = [x**2 for x in range(15)]
print(list_squares)
```

```python
list_cubes = [x**3 for x in range(15)]
print(list_cubes)
```

```python
list_nat_16 = [x for x in range(1,17)]
print(list_nat_16)
```

```python
# Using list comprehensions create list of first 25 natural numbers
# Using that list, Create another list which contains squares of the elements
# also create a list of cubes
# also create a list of square_roots of the lements in the first list
list_nat_25 = [x for x in range(1,26)]
print(list_nat_25)
```

```python
list_nat_sqs = [x**2 for x in list_nat_25]
print(list_nat_sqs)
```

```python
list_nat_cbs = [x**3 for x in list_nat_25]
print(list_nat_cbs)
```

```python
list_nat_sqrt = [x**0.5 for x in list_nat_25]
print(list_nat_sqrt)
```

```python
# from list_nat_25, let's filter the even numbers
list_25_even = [x for x in list_nat_25 if x%2==0]
print("Even numbers in the first 25 natural numbers",list_25_even)
```

```python
list_25_odd = [x for x in list_nat_25 if x%2!=0]
print("Odd numbers in the list of first 25 natural numbers",list_25_odd)
```

```python
# let's create a list any 5 countries names
countries_list = ["India","sri Lanka","BhutaN","PakisTan","nepal"]
print(countries_list)
```

```python
# Using list comprehensions can we create a list of countires in all lowercases, uppercases, Capitalized version of countries_list
countries_lower = [x.lower() for x in countries_list]
print(countries_lower)
```

```python
countries_upper = [x.upper() for x in countries_list]
print(countries_upper)
```

```python
countries_caps = [x.capitalize() for x in countries_list]
print(countries_caps)
```

### slicing and indexing

```python
print(list_nat_25)
```

```python
# fetch the element 10 from the list
list_nat_25[8]
```

```python
# Let's fetch th elements from 9 till the end
# index of 9 is 8
list_nat_25[8:25]
```

```python
# syntax is list_name[start:end]
# here start index is included but not the end index
# syntax for fetching the elements with step is list_name [start:end:step]
# let's fetch every element alternativiely from the beginning
list_nat_25[0:25:2]
```

```python
list_nat_25[0::2]
```

```python
list_nat_25[0:-1:2]
```

```python
list_nat_25[0::3]
```

```python
# fetch elements till 15 from beginnning
list_nat_25[0:15]
```

```python
# when the start position isn't mentioned it understands to begin from the first element
# when the end position isn't mentioned it understands to got till the end element
# finding squares of alternate elements in the list
list_alternate_sqs = [x**2 for x in list_nat_25[0::2]]
print(list_alternate_sqs)
```

```python
list_third_cubes = [x**3 for x in list_nat_25[0::3]]
print(list_third_cubes)
```

```python
# find the square root of every 4th element of the list from beginning to the end
list_fourth_sqrt = [x**0.5 for x in list_nat_25[0::4]]
print(list_fourth_sqrt)
```

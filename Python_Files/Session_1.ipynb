### Lets get to know the data types of the variables


```python
# Variables 'first_name' , 'last_name' and full_name
first_name = 'Kunal'
last_name = 'Rai'
# lets create variable 'full_name' using 'first_name' and 'last_name'
full_name = first_name+' '+last_name
print("My full name is",full_name)
```

### Input Method For Dynamically Creating Variables

```python
first_name = input("My First Name is")
last_name = input("My Last Name is")
full_name = first_name+' '+last_name
print("My Full Name is", full_name)
```

```python
qualification = input("My Qualification is")
print(qualification)
```

```python
age = int(input("My age is"))
print(type(age))
print("my age after 20 years will be",age+20)
```

```python
salary = float(input("My Salary is"))
expenditure = float(input("My Expenditure is"))
savings = salary - expenditure
print("My Savings are", savings)
```
